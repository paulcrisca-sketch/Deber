```python
matriz = []
filas=0
columna=0

def crear_matriz ():
    print('Arme la base de datos')
    filas = int(input('Ingrese el numero de filas: '))
    columna = int(input('Ingrese el numero de columnas: '))
    return filas,columna


def modificar_matriz(filas,columna):
    matriz.clear() 
    for i in range(filas):
        fila = []
        print(f'Ingrese lo datos de la fila {[i+1]}')

        for j in range(columna):
            dato = int(input('Ingrese informacion: '))
            fila.append(dato)
            print(f'Ingrese lo datos de la posicion {[j+1]}')
            

        matriz.append(fila)

    print(f'La matriz es {matriz}')


while True:
    print('\nIngrese la opcion a realizar')
    print('1. Crear matriz')
    print('2. definir los datos de la matriz')
    print('3. Mostrar matriz')
    print('4. Salir')

    try:
        opc = int(input('Ingrese una opcion: '))
     
        if opc == 1:
            filas,columna=crear_matriz()
            
        elif opc ==2:
            if  filas> 0 and columna > 0:
                modificar_matriz(filas,columna)
            else:
                print('Debe ingresar los datos')

        elif opc == 3:
            print(f'La matriz es {matriz}')

        elif opc == 4:
            print('Programa finalizado')
            break

        else:
            print('Opcion incorrecta')

    except ValueError:
        print('Debe ingresar un numero')
```
