```python
from dataclasses import dataclass

import re
```

```python
@ dataclass
class Logistica:
    id_rastreo:str
    item:str
    pais_origen:str
    impuesto_valor:float
    total_pagar:float
    sin_impuesto:float
```

```python
lista=[]
```

```python
for linea in data_set:
    linea_limpia=re.sub("[:;-]",",",linea).strip()
    cortado=linea_limpia.split(",")
    teclado_id=cortado[0].upper()
    numero_id=cortado[1].strip()
    id_completo=teclado_id+ numero_id
    tasa=float(cortado[-1].replace("?","").replace("#","").replace("!","").strip())
    precio=float(cortado[-3])
    calculo_de_impuesto=round(tasa*precio,2)
    total=calculo_de_impuesto+precio
    pais=cortado[-5].upper()
    producto=cortado[-7].strip().replace("_"," ").capitalize()
    
    registro= Logistica(id_completo,producto,pais,calculo_de_impuesto,total,precio)
      
    lista.append(registro)
    
    print(pais)
```

```python
print(f"{'Cedula':<10}|{'Producto':<20}|{'Pais':<10}|{'Impuesto':<10}|{'Total':<10}")

for datos in lista:
    print(f'{datos.id_rastreo:<10}|{datos.item:<20}|{datos.pais_origen:<10}|{datos.impuesto_valor:<10}|{datos.total_pagar}')
```

```python
valor_mayor=0
for e in lista:
    if e.impuesto_valor > valor_mayor:
        valor_mayor=e.impuesto_valor
        producto_indicado=e.item

print(f"El impuesto mas alto es {valor_mayor} pagado por {producto_indicado}")
```

```python
promedio=0
contador=0

for e in lista:
    promedio += e.sin_impuesto
    contador+=1
    #promedio=promedio/len(e.sin_impuesto)
promedio=round(promedio/contador,2)

print(f"El promedio de los valores sin impuestos es {promedio}")

```
